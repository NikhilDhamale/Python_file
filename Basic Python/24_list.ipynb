```python
#list indexing

marks=[67,78,89,56,76]

marks[2]=65

print(marks)
```

```python
#list slicing

print(marks[:5])
```

```python
# list method

# l.append(val)    add one element at the end


num=[4,6,7,87,98,45]

num.append(9)

print(num)

```

```python
# insert 

num.insert(3,23)
print(num)
```

```python
num.sort()
print(num)
```

```python
num.reverse()
print(num)
```

```python
#loops use in list 

nums=[3,45,68,12,23]

x=10

idx=0

for val in nums:
    if(val==x):
        print(f"{x} found at idx ={idx}")
        break
    idx+=1
```
