```python
class Car:
    def __init__(self, brand, model, fuel):
        self.brand = brand
        self.model = model
        self.fuel = fuel

    def start(self):
        print("Car started")

    def drive(self, km):
        self.fuel -= km * 0.1

        if self.fuel <= 0:
            print("Out of fuel")
            self.fuel = 0
        else:
            print("Car drove", km, "km")
            print("Remaining fuel:", self.fuel, "L")

    def show(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Fuel:", self.fuel, "L")


car = Car("Toyota", "Fortuner", 50)

car.show()
car.start()
car.drive(100)
```
