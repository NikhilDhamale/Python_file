```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def show_details(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Marks:", self.marks)

    def result(self):
        if self.marks >= 40:
            print("Result: Pass")
        else:
            print("Result: Fail")

    def grade(self):
        if self.marks >= 80:
            print("Grade: A")
        elif self.marks >= 60:
            print("Grade: B")
        elif self.marks >= 40:
            print("Grade: C")
        else:
            print("Grade: F")


student = Student("Rahul", 101, 75)

student.show_details()
student.result()
student.grade()
```
