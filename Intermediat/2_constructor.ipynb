```python
class student:
    def __init__(self,name,cgpa):
        self.name=name
        self.cgpa=cgpa

stu1=student("rahul",9.0) 
stu2=student("om",9.20)  
stu3=student("ram",8.65)        

print(stu1.cgpa)
print(stu2.name)
print(stu3.cgpa)
```

```python
class student:
    def __init__(self,name,cgpa):
        self.name=name
        self.cgpa=cgpa

    def get_cgpa(self):
        return self.cgpa

stu1=student("rahul",9.0) 
stu2=student("om",9.20)  
stu3=student("ram",8.65)        

print(stu1.cgpa)
print(stu2.name)
print(stu3.cgpa)


print(f"{stu1.name} has cgpa is : ",stu1.get_cgpa())
```
